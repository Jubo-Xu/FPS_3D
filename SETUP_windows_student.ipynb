## Part 1 — Install VS Code  👦 STUDENT

VS Code (Visual Studio Code) is the free editor we write code in.

1. Go to **https://code.visualstudio.com/**

## Part 2 — Install Anaconda  👦 STUDENT

Anaconda gives us Python **plus** a tool called `conda` that keeps each project's
packages neatly separated in an "environment". Our environment is called `pygame_env`.

1. Go to **https://www.anaconda.com/download**

2. Download the **Windows 64-bit Graphical Installer**.
3. Run it and accept the defaults, **except** watch for this screen:
   - ✅ Leave *"Register Anaconda as my default Python"* ticked.

**Check it worked:**
- Click the Windows **Start** menu and type **`Anaconda Prompt`** — you should see it appear.
- Open it. You'll get a black window with a prompt that starts with `(base)`.

> 🐍 The `(base)` at the start of the line means conda is active. Good.

> **Alternative (lighter option):** If Anaconda feels too big, **Miniforge** works

## Part 3 — Create the `pygame_env` environment  👦 STUDENT

This recreates the **exact** environment the project uses:
**Python 3.8** with `pygame`, `ursina`, `numpy`, and `Pillow`.

Open **Anaconda Prompt** and run these commands **one line at a time**.

### 3a. Create the environment (Python 3.8)
```bat
conda create -n pygame_env python=3.8
```
When it asks `Proceed ([y]/n)?`, type **`y`** and press Enter.

### 3b. Activate it
```bat
conda activate pygame_env
```
The prompt should now start with `(pygame_env)` instead of `(base)`.

### 3c. Install the game packages (pinned to the versions we use)
```bat
pip install pygame==2.6.1 ursina==5.2.0 numpy==1.24.4 Pillow==10.0.0
```
This also automatically installs the 3D engine pieces (`Panda3D`, `panda3d-gltf`, `panda3d-simplepbr`) that `ursina` needs. It may take a few minutes. ☕

> 🔊 **Bonus for the student:** On a real Windows laptop the game will actually
> play sound (the laser "pew"!). The teacher's WSL/Linux setup has no sound card,
> so this is one thing that's *better* on your machine.


### 3d. Confirm the versions match

Still in the Anaconda Prompt (with `pygame_env` active), run:
```bat
python -c "import pygame, ursina, numpy, PIL; print('pygame', pygame.__version__); print('numpy', numpy.__version__)"
```
You should see `pygame 2.6.1` and `numpy 1.24.4` printed with no errors.

> ❗ If you ever see `ModuleNotFoundError`, you're probably in the wrong environment.
> Run `conda activate pygame_env` again and check the prompt says `(pygame_env)`.

## Part 5 — Install & log in to Claude Code  👦 STUDENT

The VS Code **Claude Code** extension gives a nice chat panel, but to use the
`claude` command in the terminal we also install the small command-line tool. Both are worth having.

### 5a. Install the Claude Code command-line tool
1. Click Start → type **`PowerShell`** → open **Windows PowerShell**.
2. Paste this line and press Enter:
   ```powershell
   irm https://claude.ai/install.ps1 | iex
   ```
3. Let it finish, then **close and reopen** PowerShell so it picks up the new command.
4. Check it installed:
   ```powershell
   claude --version
   ```

> ⚠️ **Gotcha:** The `irm ... | iex` line only works in **PowerShell**. If your window
> shows `C:\>` (that's Command Prompt / CMD), use this instead:
> ```bat
> curl -fsSL https://claude.ai/install.cmd -o install.cmd && install.cmd && del install.cmd
> ```

### 5b. Log in
1. In PowerShell, type:
   ```powershell
   claude
   ```
2. A browser window opens — sign in with your Claude account.
   - You need a **Claude Pro or Max** subscription (the free plan does **not** include Claude Code).
   - Once logged in, credentials are saved — you won't have to do it again.
3. Back in VS Code, open the **Claude Code** panel from the left bar; it will use the same login.

> The VS Code extension bundles its own copy for the chat panel, so once you've
> logged in with `claude` once, the panel just works.

## Part 8 — 👨‍🏫 TEACHER: Put the project on GitHub

*This part is done **once**, by the teacher, on the machine that currently has the project
(your WSL/Linux setup). It uploads the project so the student can download it.*

You already have the GitHub CLI (`gh`) and `git` installed. Run these in the project's
**parent** folder — the one that contains `fps_3d/`.

### 8a. Decide what to share
We'll share the whole `Python-Teaching` folder (or just `fps_3d/` — your call).
The commands below assume the `fps_3d` project folder.

### 8b. Create a `.gitignore` so we don't upload junk
This keeps generated/cache files out of the repo:

### 8c. Turn the folder into a git repo and make the first commit

First log in once (opens a browser):
```bash

```
Then create the repo and push in one step:
```bash

```
- Use `--private` so it's not public. (Use `--public` if you prefer.)
- This creates `github.com/<your-username>/fps_3d` and uploads everything.

### 8e. Add the student as a collaborator
Replace `STUDENT_USERNAME` with the username the student gave you in Part 7:
```bash

```
*(On older `gh` versions use:  `gh api -X PUT repos/<your-username>/fps_3d/collaborators/STUDENT_USERNAME`)*

The student will get an email invite — they must **accept** it before they can push.

### 8f. Give the student the repo URL
It looks like:
```

```

## Part 11 — Everyday workflow: how we share changes  👦👨‍🏫 BOTH

Once set up, this is the loop you'll repeat every day. Think of it as
**"get the latest → make changes → send them back."**

### Before you start working — get the latest
```powershell
git pull
```
This downloads any changes the other person pushed. **Always pull first.**

### After you've made changes — send them
```powershell
git add .
git commit -m "Say what you changed, e.g. made bullets faster"
git push
```

### The whole cycle
```
git pull          # 1. get the latest
#   ...edit code, run the game, have fun...
git add .         # 2. stage your changes
git commit -m "…" # 3. save a labelled snapshot
git push          # 4. upload so the other person can pull it
```

> 🖱️ **Prefer buttons?** VS Code has a **Source Control** panel (the branch icon on
> the left, or `Ctrl+Shift+G`). You can type a message and click **Commit**, then use
> the **… menu → Pull / Push**. Same thing, no typing.

### If two people changed the same lines (a "merge conflict")
Don't panic — Git will mark the spot. Ask Claude Code in the panel:
*"I have a merge conflict in this file, help me resolve it."* It will walk you through it.

### Golden rules
- 🔄 **Pull before you start**, **push when you pause.** This avoids most conflicts.
- 📝 Write a short, clear commit message so the other person knows what changed.
- 🧪 Run the game after pulling to make sure everything still works.


## 🆘 Troubleshooting quick reference

| Symptom | Fix |
|---------|-----|
| `conda` not recognized | Use the **Anaconda Prompt**, not plain PowerShell — or reinstall Anaconda. |
| `(base)` won't change to `(pygame_env)` | Run `conda activate pygame_env`. If that fails, redo Part 3a. |
| `ModuleNotFoundError: pygame` / `ursina` | Wrong environment. `conda activate pygame_env` first; check the prompt. |
| VS Code runs the wrong Python | `Ctrl+Shift+P` → **Python: Select Interpreter** → pick `pygame_env`. |
| `claude` not recognized | Close & reopen the terminal after installing; or redo Part 5a in **PowerShell**. |
| `git push` says *permission denied* | Make sure you **accepted the collaborator invite** (Part 9 step 1). |
| No sound in the game | Sound needs speakers + a real Windows machine; it's normal to have none on Linux/WSL. |
| ursina window opens then closes instantly | Read the error in the terminal; often a wrong package version — redo Part 3c. |

**When in doubt:** open the **Claude Code** panel in VS Code and describe what happened —
paste any red error text. That's exactly what it's for. 🙂
